```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 999

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)



optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)




from sklearn.utils.class_weight import compute_class_weight


# =====================
# ✅ Label Smoothing Loss Function
# =====================
# Step 1: Compute class weights from training labels
from sklearn.utils.class_weight import compute_class_weight

class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_split['Emotion']),
    y=train_split['Emotion']
)

class_weights = np.power(class_weights, 0.4)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float
).to(device)



# Step 2: Update loss function to support both label smoothing and class weights


class ClassWeightedFocalLabelSmoothingLoss(nn.Module):

    def __init__(self,
                 smoothing=0.05,
                 gamma=2.0,
                 weight=None):

        super().__init__()

        self.smoothing = smoothing
        self.gamma = gamma
        self.weight = weight

    def forward(self, logits, target):

        num_classes = logits.size(1)

        with torch.no_grad():

            true_dist = torch.zeros_like(logits)

            true_dist.fill_(self.smoothing / (num_classes - 1))

            true_dist.scatter_(1,
                               target.unsqueeze(1),
                               1.0 - self.smoothing)

        log_probs = F.log_softmax(logits, dim=1)

        probs = torch.exp(log_probs)
        
        pt = probs.gather(1, target.unsqueeze(1)).squeeze(1)
        
        focal_weight = (1-pt)**self.gamma

        loss = -true_dist * log_probs

        if self.weight is not None:
            loss = loss * self.weight.unsqueeze(0)
        
        loss = loss.sum(dim=1)
        
        loss = focal_weight * loss
        
        return loss.mean()


# Step 3: Replace loss_fn with new one
loss_fn = ClassWeightedFocalLabelSmoothingLoss(
    smoothing=0.01,
    gamma=0.5,
    weight=class_weights
)
# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

# BERT-CWFLS,  seed = 123

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 123

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)




from sklearn.utils.class_weight import compute_class_weight


# =====================
# ✅ Label Smoothing Loss Function
# =====================
# Step 1: Compute class weights from training labels
from sklearn.utils.class_weight import compute_class_weight

class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_split['Emotion']),
    y=train_split['Emotion']
)

class_weights = np.power(class_weights, 0.4)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float
).to(device)



# Step 2: Update loss function to support both label smoothing and class weights


class ClassWeightedFocalLabelSmoothingLoss(nn.Module):

    def __init__(self,
                 smoothing=0.05,
                 gamma=2.0,
                 weight=None):

        super().__init__()

        self.smoothing = smoothing
        self.gamma = gamma
        self.weight = weight

    def forward(self, logits, target):

        num_classes = logits.size(1)

        with torch.no_grad():

            true_dist = torch.zeros_like(logits)

            true_dist.fill_(self.smoothing / (num_classes - 1))

            true_dist.scatter_(1,
                               target.unsqueeze(1),
                               1.0 - self.smoothing)

        log_probs = F.log_softmax(logits, dim=1)

        probs = torch.exp(log_probs)
        
        pt = probs.gather(1, target.unsqueeze(1)).squeeze(1)
        
        focal_weight = (1-pt)**self.gamma

        loss = -true_dist * log_probs

        if self.weight is not None:
            loss = loss * self.weight.unsqueeze(0)
        
        loss = loss.sum(dim=1)
        
        loss = focal_weight * loss
        
        return loss.mean()


# Step 3: Replace loss_fn with new one
loss_fn = ClassWeightedFocalLabelSmoothingLoss(
    smoothing=0.01,
    gamma=0.5,
    weight=class_weights
)
# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")


if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)
    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 42

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)




from sklearn.utils.class_weight import compute_class_weight


# =====================
# ✅ Label Smoothing Loss Function
# =====================
# Step 1: Compute class weights from training labels
from sklearn.utils.class_weight import compute_class_weight

class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_split['Emotion']),
    y=train_split['Emotion']
)

class_weights = np.power(class_weights, 0.4)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float
).to(device)



# Step 2: Update loss function to support both label smoothing and class weights


class ClassWeightedFocalLabelSmoothingLoss(nn.Module):

    def __init__(self,
                 smoothing=0.05,
                 gamma=2.0,
                 weight=None):

        super().__init__()

        self.smoothing = smoothing
        self.gamma = gamma
        self.weight = weight

    def forward(self, logits, target):

        num_classes = logits.size(1)

        with torch.no_grad():

            true_dist = torch.zeros_like(logits)

            true_dist.fill_(self.smoothing / (num_classes - 1))

            true_dist.scatter_(1,
                               target.unsqueeze(1),
                               1.0 - self.smoothing)

        log_probs = F.log_softmax(logits, dim=1)

        probs = torch.exp(log_probs)
        
        pt = probs.gather(1, target.unsqueeze(1)).squeeze(1)
        
        focal_weight = (1-pt)**self.gamma

        loss = -true_dist * log_probs

        if self.weight is not None:
            loss = loss * self.weight.unsqueeze(0)
        
        loss = loss.sum(dim=1)
        
        loss = focal_weight * loss
        
        return loss.mean()


# Step 3: Replace loss_fn with new one
loss_fn = ClassWeightedFocalLabelSmoothingLoss(
    smoothing=0.01,
    gamma=0.5,
    weight=class_weights
)
# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")


# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")
```

# Only BERT with no CWFLS, A1 seed = 999

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet

from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 999

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)










# Step 3: Replace loss_fn with new one
loss_fn = nn.CrossEntropyLoss()
# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet

from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 123

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)










# Step 3: Replace loss_fn with new one
loss_fn = nn.CrossEntropyLoss()
# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet

from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 42

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)










# Step 3: Replace loss_fn with new one
loss_fn = nn.CrossEntropyLoss()
# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 999

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)




from sklearn.utils.class_weight import compute_class_weight


# =====================
# ✅ Label Smoothing Loss Function
# =====================
# Step 1: Compute class weights from training labels
from sklearn.utils.class_weight import compute_class_weight

class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_split['Emotion']),
    y=train_split['Emotion']
)

class_weights = np.power(class_weights, 0.4)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float
).to(device)


# Step 3: Replace loss_fn with new one
loss_fn = nn.CrossEntropyLoss(
    weight=class_weights
)




# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 123

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)




from sklearn.utils.class_weight import compute_class_weight


# =====================
# ✅ Label Smoothing Loss Function
# =====================
# Step 1: Compute class weights from training labels
from sklearn.utils.class_weight import compute_class_weight

class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_split['Emotion']),
    y=train_split['Emotion']
)

class_weights = np.power(class_weights, 0.4)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float
).to(device)


# Step 3: Replace loss_fn with new one
loss_fn = nn.CrossEntropyLoss(
    weight=class_weights
)




# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 42

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)




from sklearn.utils.class_weight import compute_class_weight


# =====================
# ✅ Label Smoothing Loss Function
# =====================
# Step 1: Compute class weights from training labels
from sklearn.utils.class_weight import compute_class_weight

class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_split['Emotion']),
    y=train_split['Emotion']
)

class_weights = np.power(class_weights, 0.4)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float
).to(device)


# Step 3: Replace loss_fn with new one
loss_fn = nn.CrossEntropyLoss(
    weight=class_weights
)




# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 999

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)








# Step 2: Update loss function to support both label smoothing and class weights


class FocalLoss(nn.Module):

    def __init__(self,
                 smoothing=0.05,
                 gamma=2.0,
                 weight=None):

        super().__init__()

        self.smoothing = smoothing
        self.gamma = gamma
        self.weight = weight

    def forward(self, logits, target):

        num_classes = logits.size(1)

        # Standard one-hot targets (no label smoothing)
        true_dist = torch.zeros_like(logits)
        true_dist.scatter_(
            1,
            target.unsqueeze(1),
            1.0
        )
        log_probs = F.log_softmax(logits, dim=1)

        probs = torch.exp(log_probs)
        
        pt = probs.gather(1, target.unsqueeze(1)).squeeze(1)
        
        focal_weight = (1-pt)**self.gamma

        loss = -true_dist * log_probs

        
        loss = loss.sum(dim=1)
        
        loss = focal_weight * loss
        
        return loss.mean()


# Step 3: Replace loss_fn with new one
loss_fn = FocalLoss(
    smoothing=0.0,
    gamma=0.5,
    weight=None
)
# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 123

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)








# Step 2: Update loss function to support both label smoothing and class weights


class FocalLoss(nn.Module):

    def __init__(self,
                 smoothing=0.05,
                 gamma=2.0,
                 weight=None):

        super().__init__()

        self.smoothing = smoothing
        self.gamma = gamma
        self.weight = weight

    def forward(self, logits, target):

        num_classes = logits.size(1)

        # Standard one-hot targets (no label smoothing)
        true_dist = torch.zeros_like(logits)
        true_dist.scatter_(
            1,
            target.unsqueeze(1),
            1.0
        )
        log_probs = F.log_softmax(logits, dim=1)

        probs = torch.exp(log_probs)
        
        pt = probs.gather(1, target.unsqueeze(1)).squeeze(1)
        
        focal_weight = (1-pt)**self.gamma

        loss = -true_dist * log_probs

        
        loss = loss.sum(dim=1)
        
        loss = focal_weight * loss
        
        return loss.mean()


# Step 3: Replace loss_fn with new one
loss_fn = FocalLoss(
    smoothing=0.0,
    gamma=0.5,
    weight=None
)
# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

# Only Focal Loss, Ablation A3, seed = 42

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 123

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)








# Step 2: Update loss function to support both label smoothing and class weights


class FocalLoss(nn.Module):

    def __init__(self,
                 smoothing=0.05,
                 gamma=2.0,
                 weight=None):

        super().__init__()

        self.smoothing = smoothing
        self.gamma = gamma
        self.weight = weight

    def forward(self, logits, target):

        num_classes = logits.size(1)

        # Standard one-hot targets (no label smoothing)
        true_dist = torch.zeros_like(logits)
        true_dist.scatter_(
            1,
            target.unsqueeze(1),
            1.0
        )
        log_probs = F.log_softmax(logits, dim=1)

        probs = torch.exp(log_probs)
        
        pt = probs.gather(1, target.unsqueeze(1)).squeeze(1)
        
        focal_weight = (1-pt)**self.gamma

        loss = -true_dist * log_probs

        
        loss = loss.sum(dim=1)
        
        loss = focal_weight * loss
        
        return loss.mean()


# Step 3: Replace loss_fn with new one
loss_fn = FocalLoss(
    smoothing=0.0,
    gamma=0.5,
    weight=None
)
# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

# Only Label Smoothing, Ablation A4, seed = 999

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 999

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)




from sklearn.utils.class_weight import compute_class_weight





# Step 2: Update loss function to support both label smoothing and class weights



class LabelSmoothingLoss(nn.Module):

    def __init__(self, smoothing=0.01):
        super().__init__()
        self.smoothing = smoothing

    def forward(self, logits, target):

        num_classes = logits.size(1)

        with torch.no_grad():

            true_dist = torch.zeros_like(logits)

            true_dist.fill_(self.smoothing / (num_classes - 1))

            true_dist.scatter_(
                1,
                target.unsqueeze(1),
                1.0 - self.smoothing
            )

        log_probs = F.log_softmax(logits, dim=1)

        loss = (-true_dist * log_probs).sum(dim=1)

        return loss.mean()



# Step 3: Replace loss_fn with new one
loss_fn = LabelSmoothingLoss(
    smoothing=0.01
)


# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 123

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)




from sklearn.utils.class_weight import compute_class_weight





# Step 2: Update loss function to support both label smoothing and class weights



class LabelSmoothingLoss(nn.Module):

    def __init__(self, smoothing=0.01):
        super().__init__()
        self.smoothing = smoothing

    def forward(self, logits, target):

        num_classes = logits.size(1)

        with torch.no_grad():

            true_dist = torch.zeros_like(logits)

            true_dist.fill_(self.smoothing / (num_classes - 1))

            true_dist.scatter_(
                1,
                target.unsqueeze(1),
                1.0 - self.smoothing
            )

        log_probs = F.log_softmax(logits, dim=1)

        loss = (-true_dist * log_probs).sum(dim=1)

        return loss.mean()



# Step 3: Replace loss_fn with new one
loss_fn = LabelSmoothingLoss(
    smoothing=0.01
)


# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

# Only Label Smoothing, A4 Ablation, seed = 42

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 42

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)




from sklearn.utils.class_weight import compute_class_weight





# Step 2: Update loss function to support both label smoothing and class weights



class LabelSmoothingLoss(nn.Module):

    def __init__(self, smoothing=0.01):
        super().__init__()
        self.smoothing = smoothing

    def forward(self, logits, target):

        num_classes = logits.size(1)

        with torch.no_grad():

            true_dist = torch.zeros_like(logits)

            true_dist.fill_(self.smoothing / (num_classes - 1))

            true_dist.scatter_(
                1,
                target.unsqueeze(1),
                1.0 - self.smoothing
            )

        log_probs = F.log_softmax(logits, dim=1)

        loss = (-true_dist * log_probs).sum(dim=1)

        return loss.mean()



# Step 3: Replace loss_fn with new one
loss_fn = LabelSmoothingLoss(
    smoothing=0.01
)


# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

# only Class-weighted focal loss, ablation A5, seed = 999

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 999

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)




from sklearn.utils.class_weight import compute_class_weight


# =====================
# ✅ Label Smoothing Loss Function
# =====================
# Step 1: Compute class weights from training labels
from sklearn.utils.class_weight import compute_class_weight

class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_split['Emotion']),
    y=train_split['Emotion']
)

class_weights = np.power(class_weights, 0.4)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float
).to(device)



# Step 2: Update loss function to support both label smoothing and class weights


class ClassWeightedFocalLoss(nn.Module):
    def __init__(self,
                 gamma=0.5,
                 weight=None):
        super().__init__()

        self.gamma = gamma
        self.weight = weight

    def forward(self, logits, target):

        log_probs = F.log_softmax(logits, dim=1)
        probs = torch.exp(log_probs)

        # Probability of the true class
        pt = probs.gather(1, target.unsqueeze(1)).squeeze(1)

        # Focal weighting
        focal_weight = (1 - pt) ** self.gamma

        # Standard negative log likelihood
        ce_loss = F.nll_loss(
            log_probs,
            target,
            reduction='none',
            weight=self.weight
        )

        loss = focal_weight * ce_loss

        return loss.mean()




# Step 3: Replace loss_fn with new one
loss_fn = ClassWeightedFocalLoss(
    gamma=0.5,
    weight=class_weights
)
# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

# Only Class-Weighted Focal Loss, ablation A5, Seed = 123

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 123

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)




from sklearn.utils.class_weight import compute_class_weight


# =====================
# ✅ Label Smoothing Loss Function
# =====================
# Step 1: Compute class weights from training labels
from sklearn.utils.class_weight import compute_class_weight

class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_split['Emotion']),
    y=train_split['Emotion']
)

class_weights = np.power(class_weights, 0.4)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float
).to(device)



# Step 2: Update loss function to support both label smoothing and class weights


class ClassWeightedFocalLoss(nn.Module):
    def __init__(self,
                 gamma=0.5,
                 weight=None):
        super().__init__()

        self.gamma = gamma
        self.weight = weight

    def forward(self, logits, target):

        log_probs = F.log_softmax(logits, dim=1)
        probs = torch.exp(log_probs)

        # Probability of the true class
        pt = probs.gather(1, target.unsqueeze(1)).squeeze(1)

        # Focal weighting
        focal_weight = (1 - pt) ** self.gamma

        # Standard negative log likelihood
        ce_loss = F.nll_loss(
            log_probs,
            target,
            reduction='none',
            weight=self.weight
        )

        loss = focal_weight * ce_loss

        return loss.mean()




# Step 3: Replace loss_fn with new one
loss_fn = ClassWeightedFocalLoss(
    gamma=0.5,
    weight=class_weights
)
# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

# Only Class-Weighted Focal Loss, Ablation A5, Seed = 42

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 42

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)




from sklearn.utils.class_weight import compute_class_weight


# =====================
# ✅ Label Smoothing Loss Function
# =====================
# Step 1: Compute class weights from training labels
from sklearn.utils.class_weight import compute_class_weight

class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_split['Emotion']),
    y=train_split['Emotion']
)

class_weights = np.power(class_weights, 0.4)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float
).to(device)



# Step 2: Update loss function to support both label smoothing and class weights


class ClassWeightedFocalLoss(nn.Module):
    def __init__(self,
                 gamma=0.5,
                 weight=None):
        super().__init__()

        self.gamma = gamma
        self.weight = weight

    def forward(self, logits, target):

        log_probs = F.log_softmax(logits, dim=1)
        probs = torch.exp(log_probs)

        # Probability of the true class
        pt = probs.gather(1, target.unsqueeze(1)).squeeze(1)

        # Focal weighting
        focal_weight = (1 - pt) ** self.gamma

        # Standard negative log likelihood
        ce_loss = F.nll_loss(
            log_probs,
            target,
            reduction='none',
            weight=self.weight
        )

        loss = focal_weight * ce_loss

        return loss.mean()




# Step 3: Replace loss_fn with new one
loss_fn = ClassWeightedFocalLoss(
    gamma=0.5,
    weight=class_weights
)
# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

# Class-weight and Label Smoothing, Ablation A6, Seed = 999

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 999

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)




from sklearn.utils.class_weight import compute_class_weight


# =====================
# ✅ Label Smoothing Loss Function
# =====================
# Step 1: Compute class weights from training labels
from sklearn.utils.class_weight import compute_class_weight

class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_split['Emotion']),
    y=train_split['Emotion']
)

class_weights = np.power(class_weights, 0.4)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float
).to(device)



# Step 2: Update loss function to support both label smoothing and class weights



class ClassWeightedLabelSmoothingLoss(nn.Module):

    def __init__(self,
                 smoothing=0.01,
                 weight=None):
        super().__init__()

        self.smoothing = smoothing
        self.weight = weight

    def forward(self, logits, target):

        num_classes = logits.size(1)

        with torch.no_grad():

            true_dist = torch.zeros_like(logits)

            true_dist.fill_(self.smoothing / (num_classes - 1))

            true_dist.scatter_(
                1,
                target.unsqueeze(1),
                1.0 - self.smoothing
            )

        log_probs = F.log_softmax(logits, dim=1)

        loss = -true_dist * log_probs

        # Apply class weights
        if self.weight is not None:
            loss = loss * self.weight.unsqueeze(0)

        loss = loss.sum(dim=1)

        return loss.mean()



loss_fn = ClassWeightedLabelSmoothingLoss(
    smoothing=0.01,
    weight=class_weights
)


# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

# Class-weighted label smoothing, ablation A6, seed = 123

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 123

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)




from sklearn.utils.class_weight import compute_class_weight


# =====================
# ✅ Label Smoothing Loss Function
# =====================
# Step 1: Compute class weights from training labels
from sklearn.utils.class_weight import compute_class_weight

class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_split['Emotion']),
    y=train_split['Emotion']
)

class_weights = np.power(class_weights, 0.4)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float
).to(device)



# Step 2: Update loss function to support both label smoothing and class weights



class ClassWeightedLabelSmoothingLoss(nn.Module):

    def __init__(self,
                 smoothing=0.01,
                 weight=None):
        super().__init__()

        self.smoothing = smoothing
        self.weight = weight

    def forward(self, logits, target):

        num_classes = logits.size(1)

        with torch.no_grad():

            true_dist = torch.zeros_like(logits)

            true_dist.fill_(self.smoothing / (num_classes - 1))

            true_dist.scatter_(
                1,
                target.unsqueeze(1),
                1.0 - self.smoothing
            )

        log_probs = F.log_softmax(logits, dim=1)

        loss = -true_dist * log_probs

        # Apply class weights
        if self.weight is not None:
            loss = loss * self.weight.unsqueeze(0)

        loss = loss.sum(dim=1)

        return loss.mean()



loss_fn = ClassWeightedLabelSmoothingLoss(
    smoothing=0.01,
    weight=class_weights
)


# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

# Class-weighted Label Smmothing, Ablation A6, seed = 42

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 42

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)




from sklearn.utils.class_weight import compute_class_weight


# =====================
# ✅ Label Smoothing Loss Function
# =====================
# Step 1: Compute class weights from training labels
from sklearn.utils.class_weight import compute_class_weight

class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_split['Emotion']),
    y=train_split['Emotion']
)

class_weights = np.power(class_weights, 0.4)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float
).to(device)



# Step 2: Update loss function to support both label smoothing and class weights



class ClassWeightedLabelSmoothingLoss(nn.Module):

    def __init__(self,
                 smoothing=0.01,
                 weight=None):
        super().__init__()

        self.smoothing = smoothing
        self.weight = weight

    def forward(self, logits, target):

        num_classes = logits.size(1)

        with torch.no_grad():

            true_dist = torch.zeros_like(logits)

            true_dist.fill_(self.smoothing / (num_classes - 1))

            true_dist.scatter_(
                1,
                target.unsqueeze(1),
                1.0 - self.smoothing
            )

        log_probs = F.log_softmax(logits, dim=1)

        loss = -true_dist * log_probs

        # Apply class weights
        if self.weight is not None:
            loss = loss * self.weight.unsqueeze(0)

        loss = loss.sum(dim=1)

        return loss.mean()



loss_fn = ClassWeightedLabelSmoothingLoss(
    smoothing=0.01,
    weight=class_weights
)


# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 999

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)







# Step 2: Update loss function to support both label smoothing and class weights


class FocalLabelSmoothingLoss(nn.Module):

    def __init__(self,
                 smoothing=0.01,
                 gamma=0.5):

        super().__init__()

        self.smoothing = smoothing
        self.gamma = gamma

    def forward(self, logits, target):

        num_classes = logits.size(1)

        with torch.no_grad():

            true_dist = torch.zeros_like(logits)

            true_dist.fill_(self.smoothing / (num_classes - 1))

            true_dist.scatter_(
                1,
                target.unsqueeze(1),
                1.0 - self.smoothing
            )

        log_probs = F.log_softmax(logits, dim=1)

        probs = torch.exp(log_probs)

        pt = probs.gather(
            1,
            target.unsqueeze(1)
        ).squeeze(1)

        focal_weight = (1 - pt) ** self.gamma

        loss = -true_dist * log_probs

        # ❌ No class weights

        loss = loss.sum(dim=1)

        loss = focal_weight * loss

        return loss.mean()



loss_fn = FocalLabelSmoothingLoss(
    smoothing=0.01,
    gamma=0.5
)



# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

# Focal Loss with Label Smoothing, Ablation A7, Seed = 123

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 123

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)







# Step 2: Update loss function to support both label smoothing and class weights


class FocalLabelSmoothingLoss(nn.Module):

    def __init__(self,
                 smoothing=0.01,
                 gamma=0.5):

        super().__init__()

        self.smoothing = smoothing
        self.gamma = gamma

    def forward(self, logits, target):

        num_classes = logits.size(1)

        with torch.no_grad():

            true_dist = torch.zeros_like(logits)

            true_dist.fill_(self.smoothing / (num_classes - 1))

            true_dist.scatter_(
                1,
                target.unsqueeze(1),
                1.0 - self.smoothing
            )

        log_probs = F.log_softmax(logits, dim=1)

        probs = torch.exp(log_probs)

        pt = probs.gather(
            1,
            target.unsqueeze(1)
        ).squeeze(1)

        focal_weight = (1 - pt) ** self.gamma

        loss = -true_dist * log_probs

        # ❌ No class weights

        loss = loss.sum(dim=1)

        loss = focal_weight * loss

        return loss.mean()



loss_fn = FocalLabelSmoothingLoss(
    smoothing=0.01,
    gamma=0.5
)



# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```

# Focal loss with label smoothing, Ablation A7, seed = 42

```python
# =====================
# 📦 Install + Imports
# =====================
!pip install transformers tqdm --quiet
!pip install imbalanced-learn --quiet
import torch.nn.functional as F
from transformers import get_linear_schedule_with_warmup
from sklearn.model_selection import train_test_split
import pandas as pd
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import mean_squared_error
from sklearn.metrics import zero_one_loss
from sklearn.metrics import fowlkes_mallows_score
from sklearn.metrics import mean_absolute_error
from transformers import BertTokenizer, BertModel
from imblearn.metrics import geometric_mean_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    cohen_kappa_score,
    hamming_loss,
    jaccard_score,
    roc_auc_score,
    log_loss,
    top_k_accuracy_score,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
# Add plotting libraries here
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.notebook import tqdm
import time

# ✅ Add resource tracking
import os
import psutil
import tracemalloc


import random

seed = 42

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False



train_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/train/train_sent_emo.csv')
test_df = pd.read_csv('/kaggle/input/datasets/zaber666/meld-dataset/MELD-RAW/MELD.Raw/test_sent_emo.csv')

le = LabelEncoder()
train_df['Emotion'] = le.fit_transform(train_df['Emotion'])
test_df['Emotion'] = le.transform(test_df['Emotion'])

num_classes = len(le.classes_)   

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# ==================== DATASET CLASS ====================
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenizer(texts, truncation=True, padding=True, max_length=64, return_tensors="pt")
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __getitem__(self, idx):
        return {
            'input_ids': self.encodings['input_ids'][idx],
            'attention_mask': self.encodings['attention_mask'][idx],
            'token_type_ids': self.encodings['token_type_ids'][idx],  # ✅ Add this
            'labels': self.labels[idx]
        }

    def __len__(self):
        return len(self.labels)

# ==================== TRAIN / VALIDATION SPLIT ====================

train_split, val_split = train_test_split(
    train_df,
    test_size=0.20,              # 20% validation
    random_state=42,
    stratify=train_df["Emotion"] # Preserve class distribution
)

print(f"Training samples   : {len(train_split)}")
print(f"Validation samples : {len(val_split)}")
print(f"Test samples       : {len(test_df)}")

train_ds = TextDataset(
    train_split["Utterance"].tolist(),
    train_split["Emotion"].tolist()
)

val_ds = TextDataset(
    val_split["Utterance"].tolist(),
    val_split["Emotion"].tolist()
)

test_ds = TextDataset(
    test_df["Utterance"].tolist(),
    test_df["Emotion"].tolist()
)    

# ==================== BERT MODEL CLASS ====================
class BERTEmotionClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.bert = BertModel.from_pretrained('bert-base-uncased')
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask, token_type_ids):
        outputs = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids
        )
        pooled_output = outputs.last_hidden_state[:, 0]  # CLS token
        pooled_output = self.dropout(pooled_output)
        return self.classifier(pooled_output)

model = BERTEmotionClassifier(num_classes)

device = torch.device("cuda")
model.to(device)   


# ============================================
# Reset GPU memory statistics
# ============================================

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats(device)


g = torch.Generator()
g.manual_seed(42)

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    generator=g,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    pin_memory=True
)







optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5, weight_decay=0.01)

num_epochs = 10

total_training_steps = len(train_loader) * num_epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_training_steps),
    num_training_steps=total_training_steps
)







# Step 2: Update loss function to support both label smoothing and class weights


class FocalLabelSmoothingLoss(nn.Module):

    def __init__(self,
                 smoothing=0.01,
                 gamma=0.5):

        super().__init__()

        self.smoothing = smoothing
        self.gamma = gamma

    def forward(self, logits, target):

        num_classes = logits.size(1)

        with torch.no_grad():

            true_dist = torch.zeros_like(logits)

            true_dist.fill_(self.smoothing / (num_classes - 1))

            true_dist.scatter_(
                1,
                target.unsqueeze(1),
                1.0 - self.smoothing
            )

        log_probs = F.log_softmax(logits, dim=1)

        probs = torch.exp(log_probs)

        pt = probs.gather(
            1,
            target.unsqueeze(1)
        ).squeeze(1)

        focal_weight = (1 - pt) ** self.gamma

        loss = -true_dist * log_probs

        # ❌ No class weights

        loss = loss.sum(dim=1)

        loss = focal_weight * loss

        return loss.mean()



loss_fn = FocalLabelSmoothingLoss(
    smoothing=0.01,
    gamma=0.5
)



# Step 4: Keep rest of your training loop unchanged



train_losses = []

# ✅ Start resource tracking
print("\n📊 Starting resource tracking...\n")
process = psutil.Process(os.getpid())
tracemalloc.start()
start_cpu_percent = process.cpu_percent(interval=1)
start_mem_info = process.memory_info()



# ==================== EARLY STOPPING ====================

# num_epochs = 10
patience = 4          # Stop after 2 epochs without improvement

best_val_acc = 0.0
best_epoch = 0
counter = 0

best_model_path = "best_bert_model.pt"

best_val_loss = float("inf")
# ==================== TRAINING LOOP ====================
model.train()
start_time = time.time()
for epoch in range(num_epochs):
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ FIXED: Added this line
        labels = batch['labels'].to(device)
    
        outputs = model(input_ids, attention_mask, token_type_ids)  # ✅ FIXED: Added token_type_ids

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        pbar.set_postfix({'loss': loss.item()})

    avg_loss = total_loss / len(train_loader)
    train_losses.append(avg_loss)
    
    print(f"Epoch {epoch+1}: Average Loss = {avg_loss:.4f}")

    # ==================== VALIDATION ====================
    model.eval()
    
    val_loss = 0
    val_preds = []
    val_truths = []
    
    with torch.no_grad():
    
        for batch in val_loader:
    
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            token_type_ids = batch["token_type_ids"].to(device)
            labels = batch["labels"].to(device)
    
            outputs = model(
                input_ids,
                attention_mask,
                token_type_ids
            )
    
            loss = loss_fn(outputs, labels)
            val_loss += loss.item()
    
            predictions = torch.argmax(outputs, dim=1)
    
            val_preds.extend(predictions.cpu().numpy())
            val_truths.extend(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = accuracy_score(val_truths, val_preds)
    
    print(f"Validation Loss: {avg_val_loss:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")

    # ==================== EARLY STOPPING ====================

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss      # <-- Missing
        best_val_acc = val_acc
        best_epoch = epoch + 1
        counter = 0
    
        torch.save(model.state_dict(), best_model_path)
    
        print("✅ Best model saved.")
    
    else:
    
        counter += 1
    
        print(f"No improvement for {counter} epoch(s).")
    
        if counter >= patience:
    
            print("\n==============================")
            print("Early stopping triggered.")
            print("==============================")
            break
    
    model.train()

    print(f"✅ Epoch {epoch+1} complete — Total Loss: {total_loss:.4f}")
end_time = time.time()

# ✅ End resource tracking
end_cpu_percent = process.cpu_percent(interval=1)
end_mem_info = process.memory_info()
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"\n⏱️ Total Training Time: {(end_time - start_time):.2f} seconds")
print(f"\n🧠 Memory Usage Before Training: {start_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"🧠 Memory Usage After Training: {end_mem_info.rss / (1024 * 1024):.2f} MB")
print(f"📈 Peak Memory Used: {peak_mem / (1024 * 1024):.2f} MB")
print(f"⚙️  CPU Utilization at End: {end_cpu_percent:.2f}%")

# Plot training loss
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(train_losses)+1), train_losses, marker='o', color='purple')
plt.title('Training Loss per Epoch using bert-base-uncased Model ')
plt.xlabel('Epoch')
plt.ylabel('Total Loss')
plt.grid(True)
plt.tight_layout()
plt.show()


# ==================== LOAD BEST MODEL ====================

model.load_state_dict(torch.load(best_model_path))

print(f"\nLoaded best model from Epoch {best_epoch}")
print(f"Best Validation Accuracy = {best_val_acc:.4f}")

# ==================== EVALUATION ====================
model.eval()
eval_start = time.time()

preds = []
truths = []
probs = []
with torch.no_grad():
    pbar = tqdm(test_loader, desc="Evaluating")
    for batch in pbar:
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        token_type_ids = batch['token_type_ids'].to(device)  # ✅ ADD THIS
        labels = batch['labels'].to(device)

        outputs = model(
            input_ids,
            attention_mask,
            token_type_ids
        )
        
        prob = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(prob, dim=1)
        
        probs.extend(prob.cpu().numpy())
        preds.extend(predictions.cpu().numpy())
        truths.extend(labels.cpu().numpy())

eval_end = time.time()

probs = np.array(probs)
# ✅ Evaluation resource usage
eval_cpu_percent = process.cpu_percent(interval=1)
eval_mem_info = process.memory_info()
print(f"\n📊 [Evaluation] CPU Usage: {eval_cpu_percent:.2f}%")
print(f"📊 [Evaluation] Memory Usage: {eval_mem_info.rss / (1024 * 1024):.2f} MB")

print("✅ Evaluation Complete\n")
print(classification_report(truths, preds, target_names=le.classes_))    

# ==================== PREDICTION FUNCTION ====================
def predict_emotion(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=64)
    tokens = {k: v.to(device) for k, v in tokens.items()}
    with torch.no_grad():
        output = model(
            input_ids=tokens["input_ids"],
            attention_mask=tokens["attention_mask"],
            token_type_ids=tokens["token_type_ids"]
        )
        pred = torch.argmax(output, dim=1).item()
    return le.inverse_transform([pred])[0]

print("Predicted emotion for 'I am feeling great':", predict_emotion("I am feeling great"))
print("Predicted emotion for 'I am not feeling well emotionally':", predict_emotion("I am not feeling well emotionally"))
print("Predicted emotion for 'I am terrified':", predict_emotion("I am terrified"))
print("Predicted emotion for 'I am stunned':", predict_emotion("I am stunned"))
print("Predicted emotion for 'I am managing':", predict_emotion("I am managing"))
print("Predicted emotion for 'I am really annoyed':", predict_emotion("I am really annoyed"))
print("Predicted emotion for 'That makes me nauseous':", predict_emotion("That makes me nauseous"))   

# ==================== CONFUSION MATRIX + PLOTS ====================
cm = confusion_matrix(truths, preds)
labels = le.classes_

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for MELD Emotion Classification using bert-base-uncased Model')
plt.show()

accuracy = accuracy_score(truths, preds)
precision, recall, f1, _ = precision_recall_fscore_support(truths, preds, average=None)

metrics_df = pd.DataFrame({
    'Emotion': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1
})

metrics_df.set_index('Emotion').plot(kind='bar', figsize=(12, 6), colormap='tab10')
plt.title('Precision, Recall, and F1-score per Emotion Class using bert-base-uncased Model')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.grid(False)
plt.show()

print(f"✅ Overall Accuracy: {accuracy:.4f}")

# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
# ========================
# Sensitivity, Specificity, Efficiency + Plot
# ========================
def calculate_metrics_from_cm(cm, class_names):

    total = cm.sum()

    sensitivity_list = []
    specificity_list = []
    efficiency_list = []
    fpr_list = []
    fnr_list = []
    fdr_list = []
    npv_list = []

    for i, cls in enumerate(class_names):

        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = total - TP - FP - FN

        sensitivity = TP / (TP + FN) if (TP + FN) > 0 else 0.0
        specificity = TN / (TN + FP) if (TN + FP) > 0 else 0.0
        efficiency = (sensitivity + specificity) / 2

        npv = TN / (TN + FN) if (TN + FN) > 0 else 0
        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
        fdr = FP / (FP + TP) if (FP + TP) > 0 else 0

        sensitivity_list.append(sensitivity)
        specificity_list.append(specificity)
        efficiency_list.append(efficiency)
        npv_list.append(npv)
        fpr_list.append(fpr)
        fnr_list.append(fnr)
        fdr_list.append(fdr)

        print(f"\nClass: {cls}")
        print(f"  Sensitivity (Recall): {sensitivity:.4f}")
        print(f"  Specificity         : {specificity:.4f}")
        print(f"  Efficiency          : {efficiency:.4f}")
        print(f"  NPV                 : {npv:.4f}")
        print(f"  FPR                 : {fpr:.4f}")
        print(f"  FNR                 : {fnr:.4f}")
        print(f"  FDR                 : {fdr:.4f}")

    print("\n========== MACRO METRICS ==========")
    print(f"Macro Sensitivity : {np.mean(sensitivity_list):.4f}")
    print(f"Macro Specificity : {np.mean(specificity_list):.4f}")
    print(f"Macro Efficiency  : {np.mean(efficiency_list):.4f}")
    print(f"Macro NPV         : {np.mean(npv_list):.4f}")
    print(f"Macro FPR         : {np.mean(fpr_list):.4f}")
    print(f"Macro FNR         : {np.mean(fnr_list):.4f}")
    print(f"Macro FDR         : {np.mean(fdr_list):.4f}")

    # Plot
    x = np.arange(len(class_names))
    width = 0.25

    plt.figure(figsize=(12,6))

    plt.bar(
        x - width,
        sensitivity_list,
        width,
        label="Sensitivity"
    )

    plt.bar(
        x,
        specificity_list,
        width,
        label="Specificity"
    )

    plt.bar(
        x + width,
        efficiency_list,
        width,
        label="Efficiency"
    )

    plt.xticks(x, class_names, rotation=45)
    plt.ylabel("Score")
    plt.xlabel("Emotion Class")
    plt.title("Sensitivity, Specificity and Efficiency per Emotion Class")
    plt.ylim(0, 1.05)
    plt.legend()
    plt.tight_layout()
    plt.show()



calculate_metrics_from_cm(cm, le.classes_)     

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(x=le.inverse_transform(train_df['Emotion']), palette='coolwarm')
plt.title('Train Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.countplot(x=le.inverse_transform(test_df['Emotion']), palette='coolwarm')
plt.title('Test Set Emotion Distribution using bert-base-uncased Model')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(10, 8))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Normalized Confusion Matrix using bert-base-uncased Model')
plt.show()

per_class_accuracy = cm.diagonal() / cm.sum(axis=1)

plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=per_class_accuracy, palette='viridis')
plt.title('Per-Class Accuracy using bert-base-uncased Model')
plt.ylabel('Accuracy')
plt.ylim(0, 1.05)
plt.xticks(rotation=45)
plt.grid(False)
plt.show()


print("\n================ OVERALL METRICS ================\n")


from sklearn.utils import resample

n_bootstraps = 1000
rng = np.random.RandomState(42)

boot_scores = []

truths_np = np.array(truths)
preds_np = np.array(preds)

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    if len(np.unique(truths_np[indices])) < 2:
        continue

    score = accuracy_score(
        truths_np[indices],
        preds_np[indices]
    )

    boot_scores.append(score)

boot_scores = np.array(boot_scores)

lower = np.percentile(boot_scores, 2.5)
upper = np.percentile(boot_scores, 97.5)

print(f"Bootstrap Accuracy 95% CI : ({lower:.4f}, {upper:.4f})")



boot_f1 = []

for _ in range(n_bootstraps):

    indices = rng.randint(0, len(truths_np), len(truths_np))

    score = f1_score(
        truths_np[indices],
        preds_np[indices],
        average="macro"
    )

    boot_f1.append(score)

boot_f1 = np.array(boot_f1)

lower = np.percentile(boot_f1, 2.5)
upper = np.percentile(boot_f1, 97.5)

print(f"Bootstrap Macro-F1 95% CI : ({lower:.4f}, {upper:.4f})")






print(f"Mean Squared Error   : {mean_squared_error(truths, preds):.4f}")

print(f"Mean Absolute Error  : {mean_absolute_error(truths, preds):.4f}")

rmse = np.sqrt(mean_squared_error(truths, preds))

print(f"RMSE                 : {rmse:.4f}")

print(
    f"Weighted Cohen Kappa : "
    f"{cohen_kappa_score(truths, preds, weights='quadratic'):.4f}"
)

truths_onehot = np.eye(num_classes)[truths]

brier = np.mean(np.sum((probs - truths_onehot) ** 2, axis=1))

print(f"Brier Score          : {brier:.4f}")

print(f"Zero-One Loss        : {zero_one_loss(truths, preds):.4f}")

print(f"Fowlkes-Mallows      : {fowlkes_mallows_score(truths, preds):.4f}")

print(f"G-Mean               : {geometric_mean_score(truths, preds, average='macro'):.4f}")

print(f"Accuracy              : {accuracy_score(truths, preds):.4f}")

print(f"Balanced Accuracy     : {balanced_accuracy_score(truths, preds):.4f}")

print(f"Macro Precision       : {precision_score(truths,preds,average='macro'):.4f}")
print(f"Macro Recall          : {recall_score(truths,preds,average='macro'):.4f}")
print(f"Macro F1              : {f1_score(truths,preds,average='macro'):.4f}")

print(f"Weighted Precision    : {precision_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted Recall       : {recall_score(truths,preds,average='weighted'):.4f}")
print(f"Weighted F1           : {f1_score(truths,preds,average='weighted'):.4f}")

print(f"Micro Precision       : {precision_score(truths,preds,average='micro'):.4f}")
print(f"Micro Recall          : {recall_score(truths,preds,average='micro'):.4f}")
print(f"Micro F1              : {f1_score(truths,preds,average='micro'):.4f}")

print(f"Matthews CorrCoeff    : {matthews_corrcoef(truths,preds):.4f}")

print(f"Cohen Kappa           : {cohen_kappa_score(truths,preds):.4f}")

print(f"Hamming Loss          : {hamming_loss(truths,preds):.4f}")

print(f"Jaccard Score         : {jaccard_score(truths,preds,average='weighted'):.4f}")

print(f"Log Loss              : {log_loss(truths,probs):.4f}")

print(f"Top-2 Accuracy        : {top_k_accuracy_score(truths,probs,k=2):.4f}")

print(f"Top-3 Accuracy        : {top_k_accuracy_score(truths,probs,k=3):.4f}")

truths_bin = label_binarize(truths, classes=np.arange(num_classes))

try:
    roc_auc = roc_auc_score(
        truths_bin,
        probs,
        average="macro",
        multi_class="ovr"
    )
    print(f"Macro ROC-AUC (OvR): {roc_auc:.4f}")
except ValueError as e:
    print("ROC-AUC could not be computed:", e)

try:
    pr_auc = average_precision_score(
        truths_bin,
        probs,
        average="macro"
    )
    print(f"Macro PR-AUC: {pr_auc:.4f}")
except ValueError as e:
    print("PR-AUC could not be computed:", e)



total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)


# ======================================================
# Expected Calibration Error (ECE)
# ======================================================

def expected_calibration_error(y_true, y_prob, n_bins=15):
    """
    Computes Expected Calibration Error (ECE)

    Parameters
    ----------
    y_true : array-like
        True class labels

    y_prob : ndarray
        Predicted probabilities (N x C)

    n_bins : int
        Number of confidence bins

    Returns
    -------
    ece : float
    """

    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)

    accuracies = (predictions == y_true)

    bin_boundaries = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0

    for i in range(n_bins):

        lower = bin_boundaries[i]
        upper = bin_boundaries[i + 1]

        if i == n_bins - 1:
            mask = (confidences >= lower) & (confidences <= upper)
        else:
            mask = (confidences >= lower) & (confidences < upper)

        if np.sum(mask) > 0:

            bin_accuracy = np.mean(accuracies[mask])
            bin_confidence = np.mean(confidences[mask])

            ece += (
                np.abs(bin_accuracy - bin_confidence)
                * np.sum(mask)
                / len(y_true)
            )

    return ece


# Expected Calibration Error
ece = expected_calibration_error(
    np.array(truths),
    probs,
    n_bins=10
)



print("\n============== MODEL STATISTICS ==============")

print(f"Expected Calibration Error (ECE): {ece:.4f}")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")

num_samples = len(test_ds)

print(f"Inference Time        : {eval_end-eval_start:.2f} sec")
print(f"Samples/sec           : {num_samples/(eval_end-eval_start):.2f}")

# ============================================
# Peak GPU Memory Usage
# ============================================

if torch.cuda.is_available():
    peak_gpu_memory = torch.cuda.max_memory_allocated(device) / (1024 ** 2)

    print(f"Peak GPU Memory       : {peak_gpu_memory:.2f} MB")

```
